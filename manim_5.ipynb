```python
%%writefile scene.py
from manim import *

class RationalRepresentation(Scene):
    def construct(self):
        self.camera.background_color = "#101526"

        # 1. Title
        title = Text(
            "Rational",
            font_size=36,
            color=WHITE
        ).to_edge(UP).shift(DOWN * 0.25)

        self.play(Write(title))
        self.wait(0.5)

        # 2. Original decimal number
        decimal = MathTex(
            r"0.1",
            font_size=52
        ).move_to(UP * 1.5)

        self.play(Write(decimal))
        self.wait(0.8)

        # 3. Convert decimal to fraction
        equation = MathTex(
            r"0.1 = \frac{1}{10}",
            font_size=52
        ).move_to(UP * 1.5)

        self.play(
            Transform(decimal, equation),
            run_time=1.5
        )
        self.wait(1)

        fraction_display = MathTex(
            r"\frac{1}{10}",
            font_size=72
        ).move_to(UP * 0.2)

        self.play(
            Transform(decimal, fraction_display),
            run_time=1
        )
        self.wait(0.5)

        # 4. Separate numerator and denominator
        numerator = MathTex(
            "1",
            font_size=48,
            color=BLUE
        ).move_to(LEFT * 2.3 + DOWN * 1.0)

        denominator = MathTex(
            "10",
            font_size=48,
            color=GREEN
        ).move_to(RIGHT * 2.3 + DOWN * 1.0)

        numerator_label = Text(
            "Numerator",
            font_size=22,
            color=BLUE
        ).next_to(numerator, DOWN, buff=0.2)

        denominator_label = Text(
            "Denominator",
            font_size=22,
            color=GREEN
        ).next_to(denominator, DOWN, buff=0.2)

        self.play(
            FadeIn(numerator),
            FadeIn(denominator),
            FadeIn(numerator_label),
            FadeIn(denominator_label)
        )
        self.wait(0.8)

        # 5. Create binary storage areas
        numerator_box = Rectangle(
            width=2.0,
            height=0.8,
            stroke_color=BLUE,
            stroke_width=2
        ).move_to(LEFT * 2.3 + DOWN * 2.5)

        denominator_box = Rectangle(
            width=2.0,
            height=0.8,
            stroke_color=GREEN,
            stroke_width=2
        ).move_to(RIGHT * 2.3 + DOWN * 2.5)

        numerator_storage_label = Text(
            "Numerator",
            font_size=20,
            color=BLUE
        ).next_to(numerator_box, DOWN, buff=0.15)

        denominator_storage_label = Text(
            "Denominator",
            font_size=20,
            color=GREEN
        ).next_to(denominator_box, DOWN, buff=0.15)

        self.play(
            Create(numerator_box),
            Create(denominator_box),
            FadeIn(numerator_storage_label),
            FadeIn(denominator_storage_label)
        )
        self.wait(0.5)

        # 6. Convert decimal integers to binary
        numerator_binary = MathTex(
            "0001",
            font_size=38,
            color=BLUE
        ).move_to(numerator.get_center())

        denominator_binary = MathTex(
            "1010",
            font_size=38,
            color=GREEN
        ).move_to(denominator.get_center())

        self.play(
            Transform(numerator, numerator_binary),
            Transform(denominator, denominator_binary),
            run_time=1.5
        )
        self.wait(0.5)

        # 7. Move binary numbers into storage
        self.play(
            numerator.animate.move_to(numerator_box.get_center()),
            denominator.animate.move_to(denominator_box.get_center()),
            FadeOut(numerator_label),
            FadeOut(denominator_label),
            FadeOut(decimal),
            run_time=1.5
        )
        self.wait(1)

        # 8. Copy binary numbers from storage
        numerator_copy = numerator.copy()
        denominator_copy = denominator.copy()

        self.add(numerator_copy, denominator_copy)

        self.play(
            numerator_copy.animate.move_to(
                LEFT * 2.3 + UP * 0.3
            ),
            denominator_copy.animate.move_to(
                RIGHT * 2.3 + UP * 0.3
            ),
            run_time=1.5
        )
        self.wait(0.5)

        # 9. Convert copied binary numbers back to decimal
        numerator_decimal = MathTex(
            "1",
            font_size=48,
            color=BLUE
        ).move_to(numerator_copy.get_center())

        denominator_decimal = MathTex(
            "10",
            font_size=48,
            color=GREEN
        ).move_to(denominator_copy.get_center())

        self.play(
            Transform(numerator_copy, numerator_decimal),
            Transform(denominator_copy, denominator_decimal),
            run_time=1.5
        )
        self.wait(0.8)

        # 10. Combine into a fraction
        fraction_line = Line(
            LEFT * 0.45,
            RIGHT * 0.45,
            color=WHITE,
            stroke_width=2
        ).move_to(ORIGIN)

        self.play(
            numerator_copy.animate.move_to(UP * 0.35),
            denominator_copy.animate.move_to(DOWN * 0.35),
            Create(fraction_line),
            run_time=1.5
        )

        self.wait(2)
```
