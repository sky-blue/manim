```python

%%writefile scene.py
from manim import *

class DecimalRepresentation(Scene):
    def construct(self):
        self.camera.background_color = "#101526"

        # 1. Title
        title = Text(
            "Decimal",
            font_size=36,
            color=WHITE
        ).to_edge(UP).shift(DOWN * 0.25)

        self.play(Write(title))
        self.wait(0.5)

        # 2. Original decimal number
        decimal = MathTex(
            "0.1",
            font_size=56
        ).move_to(UP * 1.5)

        self.play(Write(decimal))
        self.wait(1)

        # 3. Convert to coefficient and exponent form
        scientific = MathTex(
            "1", r"\times", "10", "^{-1}",
            font_size=56
        ).move_to(UP * 1.5)

        self.play(
            Transform(decimal, scientific),
            run_time=1.5
        )
        self.wait(1)

        # 4. Create storage areas
        sign_box = Rectangle(
            width=1.5,
            height=0.9,
            stroke_color=BLUE,
            stroke_width=2
        ).move_to(LEFT * 3.5 + DOWN * 2.0)

        coefficient_box = Rectangle(
            width=1.5,
            height=0.9,
            stroke_color=GREEN,
            stroke_width=2
        ).move_to(DOWN * 2.0)

        exponent_box = Rectangle(
            width=1.5,
            height=0.9,
            stroke_color=GOLD,
            stroke_width=2
        ).move_to(RIGHT * 3.5 + DOWN * 2.0)

        sign_label = Text(
            "Sign",
            font_size=20,
            color=BLUE
        ).next_to(sign_box, DOWN, buff=0.15)

        coefficient_label = Text(
            "Coefficient",
            font_size=20,
            color=GREEN
        ).next_to(coefficient_box, DOWN, buff=0.15)

        exponent_label = Text(
            "Exponent",
            font_size=20,
            color=GOLD
        ).next_to(exponent_box, DOWN, buff=0.15)

        self.play(
            Create(sign_box),
            Create(coefficient_box),
            Create(exponent_box),
            FadeIn(sign_label),
            FadeIn(coefficient_label),
            FadeIn(exponent_label)
        )
        self.wait(0.5)

        # 5. Prepare coefficient and exponent copies
        coefficient_copy = scientific[0].copy()
        exponent_copy = scientific[3].copy()

        coefficient_copy.set_color(GREEN)
        exponent_copy.set_color(GOLD)

        # 6. Move values into storage
        self.add(coefficient_copy, exponent_copy)

        self.play(
            coefficient_copy.animate.move_to(
                coefficient_box.get_center()
            ),
            exponent_copy.animate.move_to(
                exponent_box.get_center()
            ),
            FadeIn(
                MathTex(
                    "+",
                    font_size=42,
                    color=BLUE
                ).move_to(sign_box.get_center())
            ),
            run_time=1.8
        )

        self.wait(1)

        # 7. Remove the original expression
        self.play(
            FadeOut(decimal),
            FadeOut(sign),
            run_time=1
        )
        self.wait(0.5)

        # 8. Copy stored values out of storage
        sign_out = sign_copy.copy()
        coefficient_out = coefficient_copy.copy()
        exponent_out = exponent_copy.copy()

        self.add(sign_out, coefficient_out, exponent_out)

        self.play(
            sign_out.animate.move_to(LEFT * 2.2 + UP * 0.5),
            coefficient_out.animate.move_to(LEFT * 0.8 + UP * 0.5),
            exponent_out.animate.move_to(RIGHT * 1.8 + UP * 0.8),
            run_time=1.5
        )
        self.wait(0.5)

        # 9. Reconstruct 1 x 10^-1
        final_expression = MathTex(
            "1", r"\times", "10", "^{-1}",
            font_size=56
        ).move_to(UP * 1.5)

        sign_out = sign_copy.copy()

        self.add(sign_out)

        self.play(
            Transform(coefficient_out, final_expression[0]),
            Transform(exponent_out, final_expression[3]),
            sign_out.animate.move_to(
                final_expression[0].get_left() + LEFT * 0.3
            ),
            FadeIn(final_expression[1]),
            FadeIn(final_expression[2]),
            run_time=1.5
        )

        self.play(
            FadeOut(sign_out),
            run_time=0.5
        )

        self.wait(1)

        # 10. Convert back to decimal
        final_decimal = MathTex(
            "0.1",
            font_size=56
        ).move_to(final_expression.get_center())

        self.play(
            FadeOut(coefficient_out),
            FadeOut(exponent_out),
            Transform(final_expression, final_decimal),
            run_time=1.5
        )

        self.wait(2)
```
